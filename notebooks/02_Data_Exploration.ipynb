```python
def convert_sqft_to_num(x):
    tokens = x.split('-')

    if len(tokens) == 2:
        return (float(tokens[0]) + float(tokens[1])) / 2

    try:
        return float(x)
    except:
        return None
```

## Feature Engineering - Step 6

### Analyze Location Frequency

**Reason:**

The dataset contains many unique locations. Some locations have only a few property listings, making statistical analysis unreliable. Locations with very few records will later be grouped into **"Other"** to improve the quality of analysis.

```python
df["location"] = df["location"].str.strip()
```

```python
location_stats = df.groupby("location")["location"].count().sort_values(ascending=False)
```

```python
location_stats.head(20)
```

```python
len(location_stats[location_stats <= 10])
```

## Feature Engineering - Step 7

### Group Rare Locations

**Reason:**

Some locations have very few property listings, making meaningful analysis difficult. These low-frequency locations are grouped into **"Other"** to reduce noise and improve statistical analysis.

```python
location_stats_less_than_10 = location_stats[location_stats <= 10]
```

```python
len(location_stats_less_than_10)
```

```python
df["location"] = df["location"].apply(
    lambda x: "Other" if x in location_stats_less_than_10 else x
)
```

# Exploratory Data Analysis (EDA)

## Objective

The goal of Exploratory Data Analysis is to understand trends, patterns, and relationships in the Bengaluru housing dataset through statistical summaries and visualizations. These insights will support business decisions and the development of the CityLife Score.

```python
plt.figure(figsize=(10,6))
plt.hist(df["price"], bins=50)

plt.title("Distribution of House Prices")
plt.xlabel("Price (Lakhs)")
plt.ylabel("Number of Properties")

plt.show()
```

### Business Insight

Most properties are concentrated in the lower price range, while a small number of luxury properties create a long right tail. This indicates that the housing market is positively skewed.

```python
df["bhk"].value_counts().sort_index().plot(kind="bar")

plt.title("Number of Properties by BHK")
plt.xlabel("BHK")
plt.ylabel("Count")

plt.show()
```

### Business Insight

2 BHK and 3 BHK properties dominate the Bengaluru housing market, suggesting they represent the highest demand among buyers.

```python
plt.savefig("../images/price_distribution.png", dpi=300, bbox_inches="tight")
```

## Exploratory Data Analysis - Step 3

### Top 10 Locations by Average Property Price

**Objective:**

Identify the locations with the highest average property prices in Bengaluru. This helps understand premium residential areas and provides valuable insights for home buyers and real estate analysts.

```python
top_locations = (
    df.groupby("location")["price"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

top_locations
```

```python
plt.figure(figsize=(12,6))

top_locations.plot(kind="bar")

plt.title("Top 10 Most Expensive Locations")
plt.xlabel("Location")
plt.ylabel("Average Price (Lakhs)")

plt.xticks(rotation=45)

plt.show()
```

### Business Insight

The chart highlights Bengaluru's premium residential locations with the highest average property prices. These areas are likely to offer better infrastructure, connectivity, and amenities, making them attractive to high-income buyers and investors.

```python
cheapest_locations = (
    df.groupby("location")["price"]
    .mean()
    .sort_values()
    .head(10)
)

cheapest_locations
```

```python
plt.figure(figsize=(12,6))

cheapest_locations.plot(kind="bar")

plt.title("Top 10 Most Affordable Locations")
plt.xlabel("Location")
plt.ylabel("Average Price (Lakhs)")

plt.xticks(rotation=45)

plt.show()
```

```python
### Business Insight

Affordable locations may be suitable for first-time home buyers, students, or budget-conscious families. However, affordability should be evaluated alongside connectivity and amenities, which will be incorporated later through the CityLife Score.
```

## Exploratory Data Analysis - Step 5

### Price Distribution by BHK

**Objective:**

Analyze how property prices vary across different BHK configurations. This helps understand market trends and buyer preferences.

```python
plt.figure(figsize=(10,6))

df.boxplot(column="price", by="bhk")

plt.title("Price Distribution by BHK")
plt.suptitle("")  # Removes automatic title
plt.xlabel("BHK")
plt.ylabel("Price (Lakhs)")

plt.show()
```

### Business Insight

Properties with higher BHK configurations generally have higher prices. However, the presence of outliers indicates that factors such as location, amenities, and luxury features also influence property prices.

## Exploratory Data Analysis - Step 6

### Total Square Feet vs Property Price

**Objective:**

Examine the relationship between property size and price to understand whether larger homes generally command higher prices.

```python
plt.figure(figsize=(10,6))

plt.scatter(df["total_sqft"], df["price"], alpha=0.5)

plt.title("Total Square Feet vs Price")
plt.xlabel("Total Square Feet")
plt.ylabel("Price (Lakhs)")

plt.show()
```

### Business Insight

There is a positive relationship between property size and price. Larger homes generally cost more, although location and other factors introduce variation.
